### **Feature Engineering**

```python
# Create a copy of the cleaned data
df = original_data.copy()

# Create Frequency feature: number of orders per customer
df['Frequency'] = df.groupby('Customer ID')['Order ID'].transform('count')

# Quick check
print(df[['Customer ID', 'Order ID', 'Frequency']].head(10))
```

##### **Recency - Feature Engineering**

- Instead of using today’s system date, I've anchored recency to the **latest delivery date in the dataset** (December 2016).  
- This ensures recency values reflect the dataset’s timeframe (2012–2016) and avoids inflated values that would occur if we compared against the actual current date (2025).  

Formula:  
\[
**Recency} = {Reference Date (max Delivery Date)} - {Customer's Delivery Date}**
\]

```python
# Reference date = last delivery date in dataset
ref_date = pd.to_datetime(df['Delivery_Date']).max()

# Calculate Recency = days since last delivery
df['Recency'] = (ref_date - pd.to_datetime(df['Delivery_Date'])).dt.days

# Quick check
print(df[['Customer ID', 'Delivery_Date', 'Recency']].head(10))
```

##### **CLV - Feature Engineering**

```python
# Convert Total Revenue from string with currency to float
df['Total Revenue'] = df['Total Revenue'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Temporary CLV column = Total Revenue
df['CLV'] = df['Total Revenue']

# Quick check
print(df[['Customer ID', 'Total Revenue', 'CLV']].head(10))
```

##### **Avg Unit Cost - Feature Engineering**

```python
# Convert Unit Cost to float if not already numeric
df['Unit Cost'] = df['Unit Cost'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# Temporary Avg Unit Cost = Unit Cost
df['Avg_Unit_Cost'] = df['Unit Cost']

# Quick check
print(df[['Customer ID', 'Unit Cost', 'Avg_Unit_Cost']].head(10))
```

##### **Age - Feature Engineering**
We calculate each customer’s age to capture demographic differences that may influence purchasing behaviour.  

- For consistency with Recency, Age is anchored to the **latest order date in the dataset (2016)** rather than today’s date.  

Formula:  
**Age = (Max Order Date – Date of Birth) ÷ 365**

```python
# Convert Birthdate to datetime
df['DOB'] = pd.to_datetime(df['Customer_BirthDate'], errors='coerce')

# Reference date = last order date in dataset
ref_date = pd.to_datetime(df['Order_Date']).max()

# Calculate Age in years
df['Age'] = (ref_date - df['DOB']).dt.days // 365

# Quick check
print(df[['Customer ID', 'Customer_BirthDate', 'Age']].head(10))
```

```python
# Check new shape after feature engineering
print("DataFrame shape:", df.shape)

print("Columns:", df.columns.tolist())
```

### **Standardization of Transformed Features**

**X_scaled = Scaled version of Customer Data**

```python
from sklearn.preprocessing import StandardScaler
# Select features
features = ['Age','Recency','Frequency','CLV','Avg_Unit_Cost']
X = customer_df[features]

# Initialize scaler
scaler = StandardScaler()

# Scale all features
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for readability
X_scaled = pd.DataFrame(X_scaled, columns=features, index=customer_df.index)

# Quick check
X_scaled.head()
```

### **K-Means Clustering**



**Selecting Optimum Value of K**
> **Elbow Method**


```python
from sklearn.cluster import KMeans

inertia = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.show()
```

###### **Elbow Method – Choosing k**

The Elbow curve drops sharply until **k = 4**, with only minor gains after **k = 5**.  
Thus, the optimal k is likely **4 or 5**.  
We will validate by running the **Silhouette Score** for **k = 2 to 7**.  

 ###  **Silhouette Method**

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Silhouette analysis for k = 2...7
silhouette_scores = []
K = range(2, 8)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"k = {k}, silhouette score = {score:.4f}")

# Plot silhouette scores
plt.figure(figsize=(8,5))
plt.plot(K, silhouette_scores, 'ro-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Method for Optimal k')
plt.show()
```

> **The peak is at k = 5 (0.2671).**
> * **So, k = 5 is the best choice:**
* 	Highest silhouette score
* 	Matches the Elbow flattening

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
from sklearn.decomposition import PCA

# Use the scaled matrix
X_mat = X_scaled.values if hasattr(X_scaled, "values") else X_scaled

# 2D projection for the right-hand scatter (visual only)
pca = PCA(n_components=2, random_state=42)
X_2d = pca.fit_transform(X_mat)

# ---- k fixed at 5 ----
n_clusters = 5
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# ----- Silhouette plot (left) -----
ax1.set_xlim([-0.1, 1])
ax1.set_ylim([0, X_mat.shape[0] + (n_clusters + 1) * 10])

kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_mat)

silhouette_avg = silhouette_score(X_mat, cluster_labels)
print(f"For n_clusters = {n_clusters}, average silhouette_score = {silhouette_avg:.4f}")

sample_silhouette_values = silhouette_samples(X_mat, cluster_labels)

y_lower = 10
for i in range(n_clusters):
    ith_cluster_sil_vals = sample_silhouette_values[cluster_labels == i]
    ith_cluster_sil_vals.sort()
    size_i = ith_cluster_sil_vals.shape[0]
    y_upper = y_lower + size_i

    color = cm.nipy_spectral(float(i) / n_clusters)
    ax1.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_sil_vals,
                      facecolor=color, edgecolor=color, alpha=0.7)

    ax1.text(-0.05, y_lower + 0.5 * size_i, str(i))
    y_lower = y_upper + 10  # space between clusters

ax1.axvline(x=silhouette_avg, color="red", linestyle="--")
ax1.set_title("Silhouette plot (k = 5)")
ax1.set_xlabel("Silhouette coefficient")
ax1.set_ylabel("Cluster label")
ax1.set_yticks([])
ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

# ----- Cluster visualization in PCA space (right) -----
colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
ax2.scatter(X_2d[:, 0], X_2d[:, 1], marker='.', s=30, lw=0, alpha=0.7,
            c=colors, edgecolor='k')

# Project centers to the same PCA space
centers_2d = pca.transform(kmeans.cluster_centers_)
ax2.scatter(centers_2d[:, 0], centers_2d[:, 1], marker='o',
            c="white", alpha=1, s=200, edgecolor='k')

for i, c in enumerate(centers_2d):
    ax2.scatter(c[0], c[1], marker=f'${i}$', alpha=1, s=50, edgecolor='k')

ax2.set_title("Clusters visualized in PCA(2D) (k = 5)")
ax2.set_xlabel("PC1")
ax2.set_ylabel("PC2")

plt.suptitle("Silhouette analysis for KMeans clustering (k = 5)",
             fontsize=14, fontweight='bold')
plt.show()
```

### **Hierarchical Clustering**


 ⚠️ **Note:** Hierarchical clustering was performed on a **subset of 20,000 samples**, since running on the full dataset would exceed memory limits and crash the notebook.

```python
X_scaled.info()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

# 1. Sample 30,000 rows for computational efficiency
X_sample = X_scaled.sample(n=30000, random_state=42)

# 2. Hierarchical clustering function (Ward linkage)
def hierarchical_clustering(X_scaled, method='ward'):
    linkage_matrix = linkage(X_scaled, method=method, metric='euclidean')
    return linkage_matrix

# 3. Compute the linkage matrix
linkage_matrix = hierarchical_clustering(X_sample)

# 4. Extract clusters (k = 5)
k = 5
clusters = fcluster(linkage_matrix, k, criterion='maxclust')

# 5. Plot dendrogram (truncated for readability)
plt.figure(figsize=(12,6))
dendrogram(
    linkage_matrix,
    truncate_mode='lastp',
    p=10,
    leaf_rotation=90,
    leaf_font_size=10
)
plt.axhline(y=130, color='red', linestyle='--', label='k ≈ 5 cut')
plt.legend()
plt.title('Hierarchical Clustering – Truncated Dendrogram (30,000 sample)')
plt.xlabel('Merged Cluster Nodes')
plt.ylabel('Linkage Distance')
plt.tight_layout()
plt.show()
```

### **Interpreting the Dendrogram**

The dendrogram alone **doesn’t strongly justify k = 5** — it appears somewhat arbitrary. **A clearer structural break occurs around k = 3**, where the hierarchy shows a sharper separation before the major linkage jump.

* This suggests the data exhibits a **multi-scale cluster structure**: *k = 3* captures broader groupings, while *k = 5* offers finer segmentation.
* For consistency with the Elbow and Silhouette analyses, we proceed with **k = 5** as the chosen solution.

```python

perplexities = [5, 15, 25]
embeddings = {}

for perp in perplexities:
    tsne = TSNE(n_components=2, perplexity=perp, n_iter=1000,
                random_state=42, learning_rate=200)
    embeddings[perp] = tsne.fit_transform(X_subset)

# ----  Plot side by side ----
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

for ax, perp in zip(axes, perplexities):
    X_tsne = embeddings[perp]
    for cluster in np.unique(y_subset):
        ax.scatter(
            X_tsne[y_subset == cluster, 0],
            X_tsne[y_subset == cluster, 1],
            label=f"Cluster {cluster}",
            s=15, alpha=0.6
        )
    ax.set_title(f"t-SNE (perplexity = {perp})")
    ax.set_xlabel("Component 1")
    ax.set_ylabel("Component 2")

# Add legend only once (bottom center)
handles, labels = axes[-1].get_legend_handles_labels()
fig.legend(handles, labels, loc="lower center", ncol=5, bbox_to_anchor=(0.5, -0.05))

plt.tight_layout()
plt.show()
```

 **t-SNE with Perplexity = 25**

After applying **t-SNE dimensionality reduction** on a subset of 10,000 customers, **perplexity = 25** provided the most balanced view, showing clear boundaries between clusters.
This aligns well with our chosen **k-means result (k = 5**), making it a suitable parameter for running on the full dataset.

## **Final Visualisation:  Applying t-SNE on full dataset (perplexity = 25)**

```python
# Run t-SNE on the full scaled dataset ---
tsne = TSNE(
    n_components=2,
    perplexity=25,
    n_iter=1000,
    learning_rate=200,
    random_state=42,
    verbose=1
)

X_tsne = tsne.fit_transform(X_scaled)

# Combine results into a DataFrame ---
tsne_df = pd.DataFrame(X_tsne, columns=['TSNE1', 'TSNE2'])
tsne_df['Cluster'] = customer_df['Cluster']

# Visualize with color-coded clusters ---
plt.figure(figsize=(10, 8))
sns.scatterplot(
    data=tsne_df,
    x='TSNE1', y='TSNE2',
    hue='Cluster',
    palette='Set1',
    alpha=0.6,
    s=25,
    edgecolor=None
)

plt.title("t-SNE Visualization (Perplexity = 25)", fontsize=14)
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.legend(title="Cluster", loc='best')
plt.tight_layout()
plt.show()
```
